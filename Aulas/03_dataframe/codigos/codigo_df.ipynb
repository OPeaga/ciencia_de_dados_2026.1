```python
import pandas as pd 
```

```python
dados = {
    "ID": list(range(1, 51)),
    "Produto": [
        "Notebook","Mouse","Teclado","Monitor","Cadeira","Mesa","Impressora","Headset","Webcam","SSD"
    ] * 5,
    "Categoria": [
        "Informática","Informática","Informática","Informática","Móveis",
        "Móveis","Informática","Informática","Informática","Informática"
    ] * 5,
    "Preco": [
        3500,80,150,1200,850,1200,900,250,180,320,
        3800,75,170,1350,900,1300,950,280,200,350,
        3600,90,160,1250,870,1280,920,260,190,340,
        3700,85,175,1400,920,1350,980,300,210,360,
        3900,95,180,1450,950,1400,1000,320,220,380
    ],
    "Quantidade": [
        2,10,5,3,4,2,1,8,6,7,
        1,15,4,2,3,2,1,6,5,8,
        2,12,6,3,4,2,1,7,6,5,
        1,14,3,2,2,1,1,5,4,6,
        1,10,5,2,3,1,1,4,3,5
    ]
}

df = pd.DataFrame(dados)

df
```

# Método head()

O método head() mostra as primeiras linhas do DataFrame.

É útil quando estamos trabalhando com tabelas muito grandes e queremos visualizar apenas uma amostra inicial.

ou se quiser ver as primeiras 10 linhas - df.head(10)

```python
df.head(15)
```

# Método tail()

O método tail() mostra as últimas linhas do DataFrame.

```python
df.tail(15)
```

# Método info()

O método info() exibe informações gerais sobre o DataFrame.

Podemos visualizar:

- Quantidade de linhas
- Quantidade de colunas
- Tipos de dados
- Valores não nulos

```python
df.info()
```

# Método describe()

O método describe() gera estatísticas descritivas para as colunas numéricas.

```python
df.describe()
```

## Entendendo o resultado

count → quantidade de registros

mean → média dos valores

std → desvio padrão

min → menor valor

25% → primeiro quartil

50% → mediana

75% → terceiro quartil

max → maior valor

```python
df["Preco"].describe()

# Agora estamos analisando apenas a coluna Preco.
# Isso é útil quando queremos estudar uma variável específica.
```

```python
df["Quantidade"].describe()
```
