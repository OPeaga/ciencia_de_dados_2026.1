```python
# Importando a biblioteca Pandas
# O apelido 'pd' é o mais utilizado pela comunidade
import pandas as pd
```

```python
# Criando uma Series
# Uma Series é parecida com uma coluna de uma planilha
# Cada valor recebe automaticamente um índice (0,1,2,3...)
nomes = pd.Series(['Ana','Pedro','Rian','Luiz','Carla'])

# Exibindo a Series
print(nomes)
```

```python
# Acessando elementos da Series
# O índice 3 corresponde ao quarto elemento
print(nomes[3])
```

```python
# Criando índices personalizados
# Em vez de usar 0,1,2 podemos usar códigos personalizados
alunos = pd.Series(
    ['João','Maria','Lucas'],
    index=['A1','A2','A3']
)

print(alunos)
```

```python
# Acessando pelo índice personalizado
print(alunos['A2'])
```
